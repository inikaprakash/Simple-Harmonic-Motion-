# Complex Numbers and Simple Harmonic Motion

Use the sliders below to explore amplitude, angular frequency, phase and damping. The complex trajectory is shown together with the real displacement.

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# Complex damped harmonic motion:
# z(t) = A exp(-gamma t) exp(i(omega t + phi))
def update_plots(amplitude, frequency, phase_shift, damping):
    num_periods = 3
    num_points = 1000
    t = np.linspace(0, num_periods * 2 * np.pi, num_points)

    a = amplitude * np.exp(-damping * t) * np.cos(frequency * t + phase_shift)
    b = amplitude * np.exp(-damping * t) * np.sin(frequency * t + phase_shift)

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    axes[0].plot(a, b)
    axes[0].set_title('Complex Plane (Imaginary vs. Real)')
    axes[0].set_xlabel('Real Part (a)')
    axes[0].set_ylabel('Imaginary Part (b)')
    axes[0].grid(True)
    axes[0].axis('equal')

    radians_as_fraction = (frequency * t + phase_shift) / np.pi
    axes[1].plot(radians_as_fraction, a)
    axes[1].set_title('Displacement vs. Radians as Fraction of π')
    axes[1].set_xlabel('Radians (π)')
    axes[1].set_ylabel('Displacement (a)')
    axes[1].grid(True)

    plt.tight_layout()
    plt.show()

amplitude_slider = widgets.FloatSlider(
    value=1.0, min=0.1, max=2.0, step=0.1,
    description='Amplitude:', continuous_update=False
)
frequency_slider = widgets.FloatSlider(
    value=1.0, min=0.1, max=5.0, step=0.1,
    description='Frequency:', continuous_update=False
)
phase_shift_slider = widgets.FloatSlider(
    value=0.0, min=0.0, max=2 * np.pi, step=0.1,
    description='Phase Shift:', continuous_update=False
)
damping_slider = widgets.FloatSlider(
    value=0.1, min=0.0, max=1.0, step=0.05,
    description='Damping:', continuous_update=False
)

controls = widgets.VBox([
    widgets.HTML('<b>Adjustable Parameters</b>'),
    amplitude_slider,
    widgets.Label('Amplitude: maximum displacement from equilibrium.'),
    frequency_slider,
    widgets.Label('Frequency: angular-frequency parameter controlling oscillation.'),
    phase_shift_slider,
    widgets.Label('Phase Shift: initial phase angle in radians.'),
    damping_slider,
    widgets.Label('Damping: controls exponential decay.'),
])

output = widgets.Output()

def redraw(*args):
    with output:
        clear_output(wait=True)
        update_plots(
            amplitude_slider.value,
            frequency_slider.value,
            phase_shift_slider.value,
            damping_slider.value,
        )

for slider in (
    amplitude_slider, frequency_slider,
    phase_shift_slider, damping_slider
):
    slider.observe(redraw, names='value')

display(controls, output)
redraw()
```
